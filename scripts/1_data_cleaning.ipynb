```r
# load libraries
# load libraries
library(readr)
library(dplyr)
library(tidyr)
```

```r
# read in raw data
temps_raw <- read_csv("raw_data/stream_temp_raw.csv")
stations_raw <- read_csv("raw_data/station_info_raw.csv")
```

```r
temps_raw
```

```r
stations_raw
```

```r
# pivoting temp data to be tidy
temps_long <- temps_raw |>
  pivot_longer(
    cols = jan:dec,
    names_to = "month",
    values_to = "water_temp_c"
  )
temps_long
```

```r
# join my two dataset
stations_clean <- stations_raw |>
  rename_with(tolower) |>
  rename(station_id = station)

stations_clean
```

```r
steam_clean <- left_join(temps_long, stations_clean, by = "station_id")
steam_clean
```

```r
write_csv(steam_clean, "clean_data/stream_temp_clean.csv")
```
